```python
import mlflow
mlflow.set_tracking_uri("sqlite:///mlflow.db")
mlflow.set_experiment("meus_estudos")
import pandas as pd
```

```python
with mlflow.start_run(run_name = "my_run"):
    mlflow.log_metric("accuracy", 0.95)
    mlflow.log_param("custo_reais", 1000)
    mlflow.log_param("tempo_resposta", 1.2)
    mlflow.log_metric("tempo_resposta", 1.2)

print("Run completed and logged to MLflow.")
```
